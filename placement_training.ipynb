```python
N=int(input("Enter max no. of candies:"))
K=int(input("Enter min no. of candies:"))
C=1
R=N
while(C==1):
  M=int(input("Enter amout of candies needed:"))
  if(M>=N):
    print("invalid")
  else:
    R=R-M
    if(R<=K):
      R=N
    print("SOLD=",M)
    print("Remaining=",R)
    C=int(input("DO you want to continue(1or0):"))
```

```python
n =int(input("enter number of elements:"))
arr =list(map(int,input("enter the elements:").split()))
largest = second = float('-inf')
for x in arr:
  if x > largest:
    second = largest
    largest = x
  elif x > second and x != largest:
    second = x
print("second largest element:",second)
```
