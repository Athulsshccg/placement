```python
n=int(input("Enter total no. of monkeys:"))
k=int(input("Enter no. of banana monkey can eat:"))
j=int(input("Enter no. of peanut monkey can eat:"))
m=int(input("Enter total no. of banana:"))
p=int(input("Enter total no. of peanut:"))
if(k==0 or j==0):
  print("INVALID INPUT")
else:
  n=n-(m//k+p//j)
  if(m%k!=0 or p%k!=0):
    n=n-1
  print("No. of monkeys in tree:",n)
```

```python
a=int(input("enter the number first "))
b=int(input("enter the number second "))
print ("condition using + and -")
a=a+b
b=a-b
a=a-b
print ("swap 1 a =",a)
print ("swap 1 b =",b)
print ("condition using xor")
a=a^b
b=a^b
a=a^b
print ("swap 2 a =",a)
print ("swap 2 b =",b)
print ("condition using ,")
a,b=b,a
print ("swap 3 a =",a)
print ("swap 3 b =",b)
print ("condition using temp")
temp=a
a=b
b=temp
print ("swap 4 a =",a)
print ("swap 4 b =",b)
```

```python
N=int(input("Enter max no. of candies:"))
K=int(input("Enter min no. of candies:"))
C=1
R=N
while(C==1):
  M=int(input("Enter amout of candies needed:"))
  if(M>=N):
    print("invalid")
  else:
    R=R-M
    if(R<=K):
      R=N
    print("SOLD=",M)
    print("Remaining=",R)
    C=int(input("DO you want to continue(1or0):"))
```
